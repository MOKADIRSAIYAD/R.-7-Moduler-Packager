## 1. Setup — package import

```python
import os, sys
sys.path.insert(0, os.getcwd())     # current folder ko import path me jodo

import toolkit
print("Package version:", toolkit.__version__)
print("Modules in package:", toolkit.__all__)
```

## 2. Datetime and Time Module

```python
from toolkit import datetime_utils as dtu

print("Current Date and Time:", dtu.current_datetime())
print("Difference:", dtu.date_difference("2024-12-25", "2025-01-04"))
print("Custom format:", dtu.format_date("2025-01-04", "%d %B %Y (%A)"))

watch = dtu.Stopwatch()
watch.start()
sum(range(1_000_000))
print("Stopwatch:", dtu.Stopwatch.format_elapsed(watch.stop()))

dtu.countdown(3, lambda r: print("Countdown:", r))
print("Time's up!")
```

## 3. Math Module

```python
from toolkit import math_utils as mu

print("Factorial of 5:", mu.factorial(5))
print("Compound Amount:", round(mu.compound_amount(1000, 5, 2), 2))
print("Compound Interest:", round(mu.compound_interest(1000, 5, 2), 2))
print("Trigonometry (30 deg):", mu.trig_values(30))
print("log10(100):", mu.logarithm(100, 10))
print("Area of circle (r=7):", round(mu.area_circle(7), 2))
print("36.6 C in F:", round(mu.convert_units("5", 36.6), 2))
```

## 4. Random Module

```python
from toolkit import random_utils as ru

print("Random number :", ru.random_number(1, 100))
print("Random list   :", ru.random_list(5, 1, 50))
print("Password      :", ru.random_password(8))
print("OTP           :", ru.random_otp(6))
print("Sample        :", ru.random_sample(["red", "green", "blue", "yellow"], 2))
rounds, player, computer = ru.dice_game(3)
print("Dice game     :", rounds, "-> Player", player, "| Computer", computer)
```

## 5. UUID Module

```python
from toolkit import uuid_utils as uu

print("UUID4   :", uu.generate_uuid4())
print("UUID1   :", uu.generate_uuid1())
print("UUID5   :", uu.generate_uuid5("example.com"))
print("Invoice :", uu.generate_invoice_id())
```

## 6. File Operations (Custom Module)
Files `output_files/` folder me banti hain.

```python
from toolkit import file_operations as fo

print(fo.create_file("notebook_demo.txt"))
print(fo.write_file("notebook_demo.txt", "This is a sample file.\n"))
print(fo.append_file("notebook_demo.txt", "Appended line.\n"))
print(fo.read_file("notebook_demo.txt"))
print(fo.create_file("notebook_demo.txt"))     # already exists -> error handle hota hai
print(fo.read_file("missing.txt"))             # not found -> error handle hota hai
```

## 7. Explore Module Attributes using `dir()`

```python
from toolkit import explorer

for name in ("math", "toolkit.math_utils"):
    module_name, attrs = explorer.explore_module(name)
    print(f"Available Attributes in {module_name} module:")
    print(explorer.format_attributes(attrs))
    print()
```
